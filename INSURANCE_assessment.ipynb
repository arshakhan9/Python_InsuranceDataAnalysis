Code for Exploratory Data Analysis (EDA):
**Count plot for categorical columns:**
A count plot is useful for visualizing the frequency distribution of a categorical variable.

**Scatter plot for numerical columns:**
A scatter plot helps visualize the relationship between a numerical feature and the target variable (also numerical).

```python
# 4.	Explore the relationship between the feature and target column using a count plot of categorical columns and a scatter plot of numerical columns""
import matplotlib.pyplot as plt
import seaborn as sns

# Set the figure size
plt.figure(figsize=(12, 6))

# Count plot for categorical features
categorical_columns = ['sex', 'smoker', 'region']

for col in categorical_columns:
    plt.figure(figsize=(8, 6))
    sns.countplot(data=df, x=col, hue='charges')  # 'Rating' is the target variable
    plt.title(f"Count plot of {col} vs charges")
    plt.xticks(rotation=45)
    plt.show()

# Scatter plot for numerical features
numerical_columns = ['age', 'bmi', 'children', 'charges']

for col in numerical_columns:
    plt.figure(figsize=(8, 6))
    sns.scatterplot(data=df, x=col, y='charges')  # 'Rating' is the target variable
    plt.title(f"Scatter plot of {col} vs charges")
    plt.show()
```

**Visualization**

```python
# Pairplot to visualize relationships
sns.pairplot(df, hue="smoker", palette="husl")
plt.show()
```
